# Derivadas e Taxa de Variação

A derivada de uma função $f$ em um ponto $x$ corresponde a taxa instantânea de variação no referido ponto. Seu cálculo é dado pela seguinte expressão:

$\frac{d}{dx} f(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}$

```python
'''
Função que calcula a derivada para uma função "f" em um ponto "x".
'''
def df_dx(f, x):
    h = 1e-12   # define um h positivo bem pequeno (próximo de zero)
    return (f(x + h) - f(x)) / h
```

Numa intepretação geométrica, a derivada representa a inclinação da reta tangente à curva da função $f$ no ponto $x$. Seja a forma ponto-inclinação da equação da reta dada por:

$y - y_0 = m (x - x_0)$

onde $m$ é a derivada da função $f$ com relação a $x$.

```python
'''
Função que calcula o "y" da função da reta tangente à função "f" no ponto "x0".
'''
def calcular_reta_tangente(f, x, x0):
    return df_dx(f, x0) * (x - x0) + f(x0)
```

Abaixo, um exemplo gráfico da intepretação geométrica da derivada para a função $f(x) = x^2 - 2x + 1$ para $x = 1$.

```python
def fn(x):
    return x**2 - 2*x + 1

x = np.linspace(-1, 3, 100)   # valores de "x" para plotar a função "f"
x0 = 1                        # ponto para o qual se deseja verificar a reta tangente

reta_tangente = calcular_reta_tangente(fn, x, x0)

plt.plot(x, fn(x), color=cDesertSun, label='$f(x) = x^2 - 2x + 1$')
plt.plot(x, reta_tangente, color=cDarkBlue, linestyle='dashed', label=f'Tangente em $x = {x0}$')

plt.legend(facecolor=cIvory)
plt.xlabel('$x$')
plt.ylabel('$f(x)$')

_ = plt.show()
```

## Regras de Derivação

Para a execução das regras de derivação apresentadas abaixo, execute a caixa de código a seguir.

```python
'''
Função que calcula a derivada para uma função "f" em um ponto "x".
'''
def df_dx(f, x):
    h = 1e-12   # define um h positivo bem pequeno (próximo de zero)
    return (f(x + h) - f(x)) / h

'''
Função que calcula o "y" da função da reta tangente à função "f" no ponto "x0".
'''
def calcular_reta_tangente(f, x, x0):
    return df_dx(f, x0) * (x - x0) + f(x0)

'''
Função que plota um gráfico combinando a função "f" com a reta tangente ao ponto "x0".
'''
def plotar_reta_tangente(x, y, x0, reta_tangente):
    plt.plot(x, y, color=cDesertSun, label='$f(x)$')
    plt.plot(x, reta_tangente, color=cDarkBlue, linestyle='dashed', label=f'Tangente em $x = {x0}$')
    plt.legend(facecolor=cIvory)
    plt.xlabel('$x$')
    plt.ylabel('$f(x)$')
    _ = plt.show()
```

### Derivada de uma constante

```python
def f(x):
    return np.full_like(x, 2)       # retorna um vetor com a mesma dimensão de "x", contendo um valor constante (2)

x = np.linspace(-1, 3, 100)
y = f(x)
x0 = 1
reta_tangente = calcular_reta_tangente(f, x, x0)

print(f"df/dx = {df_dx(f, x0)}")

plotar_reta_tangente(x, y, x0, reta_tangente)
```

### Regra da Potência

```python
def f(x):
    return 2 * x**2 - 3 * x + 2

x = np.linspace(-1, 3, 100)
y = f(x)
x0 = 1
reta_tangente = calcular_reta_tangente(f, x, x0)

print(f"df/dx = {df_dx(f, x0)}")

plotar_reta_tangente(x, y, x0, reta_tangente)
```

### Regra da Multiplicação por Constante

```python
def f(x):
    return x**2 - 5

c = 3               # valor da constante que será multiplicada dentro e fora da função
def c_f(x):
    return c * (f(x))

x = np.linspace(-1, 3, 100)
y = c_f(x)
x0 = 1
reta_tangente = calcular_reta_tangente(c_f, x, x0)

print(f"  df/dx = {df_dx(c_f, x0)}")
print(f"c df/dx = {c * df_dx(f, x0)}")

plotar_reta_tangente(x, y, x0, reta_tangente)
```

### Regra da Soma

```python
def f(x):
    return x**2 + 2
def g(x):
    return 2 * x - 1
def sum_fg(x):
    return f(x) + g(x)

x = np.linspace(-1, 3, 100)
y = sum_fg(x)
x0 = 1
reta_tangente = calcular_reta_tangente(sum_fg, x, x0)

print(f" d/dx (f + g) = {df_dx(sum_fg, x0)}")
print(f"df/dx + dg/dx = {df_dx(f, x0) + df_dx(g, x0)}")

plotar_reta_tangente(x, y, x0, reta_tangente)
```

### Regra da Subtração

```python
def f(x):
    return x**2 - 3
def g(x):
    return 3 * x + 1
def sub_fg(x):
    return f(x) - g(x)

x = np.linspace(-1, 3, 100)
y = sub_fg(x)
x0 = 1
reta_tangente = calcular_reta_tangente(sub_fg, x, x0)

print(f" d/dx (f - g) = {df_dx(sub_fg, x0)}")
print(f"df/dx - dg/dx = {df_dx(f, x0) - df_dx(g, x0)}")

plotar_reta_tangente(x, y, x0, reta_tangente)
```

### Regra do Produto

```python
def f(x):
    return 5 * x - 2
def g(x):
    return 2 * x**2 + 3
def prod_fg(x):
    return f(x) * g(x)

x = np.linspace(-1, 3, 100)
y = prod_fg(x)
x0 = 1
reta_tangente = calcular_reta_tangente(prod_fg, x, x0)

print(f"     d/dx (f * g) = {df_dx(prod_fg, x0)}")
print(f"f dg/dx + g df/dx = {f(x0) * df_dx(g, x0) +  g(x0) * df_dx(f, x0)}")

plotar_reta_tangente(x, y, x0, reta_tangente)
```

### Regra do Quociente

```python
def f(x):
    return 3 * x - 3
def g(x):
    return x**2 + 1
def quo_fg(x):
    return f(x) / g(x)

x = np.linspace(-1, 3, 100)
y = quo_fg(x)
x0 = 1
reta_tangente = calcular_reta_tangente(quo_fg, x, x0)

print(f"             d/dx (f / g) = {df_dx(quo_fg, x0)}")
print(f"(g df/dx - f dg/dx) / g^2 = {(g(x0) * df_dx(f, x0) - f(x0) * df_dx(g, x0)) / g(x0)**2}")

plotar_reta_tangente(x, y, x0, reta_tangente)
```

### Regra da Cadeia

```python
def f(x):
    return x**2 + 3
def g(x):
    return -1 * x**2 + 4
def chain_fg(x):
    return f(g(x))

x = np.linspace(-1, 3, 100)
y = chain_fg(x)
x0 = 1
reta_tangente = calcular_reta_tangente(chain_fg, x, x0)

print(f"        d/dx (f(g)) = {df_dx(chain_fg, x0)}")
print(f"d/dg (f(g)) * dg/dx = {df_dx(f, g(x0)) * df_dx(g, x0)}")

plotar_reta_tangente(x, y, x0, reta_tangente)
```

# Derivadas Parciais

Para uma função de duas variáveis $f(x, y)$, a derivada parcial com relação a $x$ é obtida derivando a função em relação a $x$ enquanto $y$ é tratado como uma constante:

$\frac{\partial f}{\partial x} = \lim_{t \to 0} \frac{f(x + h, y) - f(x, y)}{h}$

Analogamente, a derivada parcial com relação a $y$ é dada por:

$\frac{\partial f}{\partial y} = \lim_{t \to 0} \frac{f(x, y + h) - f(x, y)}{h}$

```python
'''
Função que calcula a derivada parcial para uma função "f" com relação a "x".
'''
def df_dx(f, x, y):
    h = 1e-12       # define um h positivo bem pequeno (próximo de zero)
    return (f(x + h, y) - f(x, y)) / h

'''
Função que calcula a derivada parcial para uma função "f" com relação a "y".
'''
def df_dy(f, x, y):
    h = 1e-12       # define um h positivo bem pequeno (próximo de zero)
    return (f(x, y + h) - f(x, y)) / h
```

As derivadas parciais de funções com duas variáveis podem ser interpretadas geometricamente por meio de um plano tangente a um ponto escolhido na superfície da função. A equação do plano tangente é:

$z - z_0 = \frac{\partial}{\partial x} f(x_0, y_0) (x - x_0) + \frac{\partial}{\partial y} f(x_0, y_0) (y - y_0)$

```python
'''
Função que calcula o plano tangente a função "f" no ponto "x0", "y0"
'''
def calcular_plano_tangente(f, x, x0, y, y0):
    return df_dx(f, x0, y0) * (x - x0) + df_dy(f, x0, y0) * (y - y0) + f(x0, y0)
```

A seguir, um exemplo gráfico de uma função $f(x, y) = -2x^2 - y^2$, com um plano tangente ao ponto $x = 2$, $y = -2$ e (consequentemente) $z = -12$.

```python
def f(x, y):
    return -2*(x**2) - (y**2)

x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
x0 = 2
y0 = -2
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

plano_tangente = calcular_plano_tangente(f, X, x0, Y, y0)

print(f'df/dx = {df_dx(f, x0, y0)}')
print(f'df/dy = {df_dy(f, x0, y0)}')

size_prop = 1
fig = plt.figure(figsize=(16*size_prop, 9*size_prop))
ax = fig.add_subplot(projection='3d')

surf = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.9)
fig.colorbar(surf, shrink=0.85)
ax.plot_surface(X, Y, plano_tangente, cmap='Greys', alpha=0.7)

ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_zlabel('$f(x, y)$')

ax.zaxis.labelpad=10

_ = plt.show()
```

# Derivadas Direcionais e Vetor Gradiente

Abaixo, você pode observar o gráfico de contorno da função $f(x, y) = x^2 + y^2$, com vetores gradientes.

```python
def f(x, y):
    return x**2 + y**2

def grad_f(x, y):
    return 2*x, 2*y

x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

xp = np.linspace(-3, 3, 15)
yp = np.linspace(-3, 3, 15)
XP, YP = np.meshgrid(xp, yp)
U, V = grad_f(XP, YP)

contorno = plt.contour(X, Y, Z, levels=25, cmap='viridis')
plt.clabel(contorno, inline=True, fontsize=8)
plt.quiver(XP, YP, U, V, color='red', angles='xy', scale_units='xy', scale=10)

plt.xlabel('$x$')
plt.ylabel('$y$')

_ = plt.show()
```

# Derivadas de Ordem Dois

Se $f$ é uma função diferenciável, então sua dervidada $f'$ também é uma função que pode, por sua vez, ter sua própria derivada, isto é, $(f')' = f''$. Considere a função $f(x) = x^3 + x$, cujo gráfico para o intervalo $x \in [-3, 3]$ é apresentado abaixo.

```python
def f(x):
    return x**3 + x

x = np.linspace(-3, 3, 50)

plt.plot(x, f(x), color=cDesertSun)

plt.xlabel('$x$')
plt.ylabel('$f(x) = x^3 + x$')

_ = plt.show()
```

A derivada dessa função é $f'(x) = 3x + 1$ e representa a taxa de variação da função $f(x)$ em cada ponto do intervalo. Observe abaixo que o gráfico da derivada $f'(x)$ é uma reta.

```python
def df(x):
    return 3*x + 1

x = np.linspace(-3, 3, 50)

plt.plot(x, df(x), color=cDesertSun)

plt.xlabel('$x$')
plt.ylabel("$f'(x) = 3x + 1$")

_ = plt.show()
```

Se tormarmos quisermos verificar a taxa de variação desta taxa de variação, precisamos calcular a segunda derivada, que neste caso é $f''(x) = 3$. Isto é, uma constante, conforme demonstrado também no gráfico abaixo.

```python
def ddf(x):
    return np.full_like(x, 3) 

x = np.linspace(-3, 3, 50)

plt.plot(x, ddf(x), color=cDesertSun)

plt.xlabel('$x$')
plt.ylabel("$f''(x) = 3$")

_ = plt.show()
```
